# Bagging(same model)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.datasets import load_digits
from sklearn.metrics import accuracy_score
```

```python
digits = load_digits()
x_train , x_test , y_train , y_test = train_test_split(digits.data , digits.target , test_size = 0.2 , random_state = 42)
```

```python
bag_clf = BaggingClassifier(
             base_estimator = DecisionTreeClassifier() ,   
             n_estimators = 500 , 
             bootstrap = True , 
             n_jobs = -1
)
```

```python
bag_clf.fit(x_train , y_train)
```

```python
y_pred = bag_clf.predict(x_test)
y_pred
```

```python
accuracy = accuracy_score(y_test , y_pred)
accuracy
```

# Bagging(different model)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.datasets import load_digits
from sklearn.linear_model import LogisticRegression 
from sklearn.neighbors import KNeighborsClassifier
#from sklearn.metrics import accuracy_score
```

```python
digits = load_digits()
x_train , x_test , y_train , y_test = train_test_split(digits.data , digits.target , test_size = 0.2 , random_state = 42)
```

```python
tree_clf = DecisionTreeClassifier()
logis_clf = LogisticRegression()
knn_clf = KNeighborsClassifier
```

```python
tree_bagging = BaggingClassifier(
    estimator = tree_clf,
    n_estimators = 500,
    bootstrap = True,
    random_state = 42
)

logis_bagging = BaggingClassifier(
    estimator = logis_clf,
    n_estimators = 500,
    bootstrap = True,
    random_state = 42
)

knn_bagging = BaggingClassifier(
    estimator = knn_clf,
    n_estimators = 500,
    bootstrap = True,
    random_state = 42
)  
```

```python
# tree
tree_bagging.fit(x_train , y_train)
tree_pred = tree_bagging.predict(x_test)

#logistic
logis_bagging.fit(x_train , y_train)
logis_pred = tree_bagging.predict(x_test)

#knn
knn_bagging.fit(x_train , y_train)
knn_pred = tree_bagging.predict(x_test)
```

```python
# prediction
tree_pred = tree_bagging.predict(x_test)
logis_pred = logis_bagging.predict(x_test)
knn_pred  = knn_bagging.predict(x_test)

# print results
print(tree_pred)
print('*****************')
print(logis_pred)
print('*****************')
print(knn_pred)
```

```python
final_pred = np.round((tree_pred  +logis_pred  +knn_pred) / 3)
final_pred
```

```python
accuracy = accuracy_score(y_test , final_pred)
accuracy
```

# Bagging(Random_Forest)

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
digits = load_digits()
x_train , x_test , y_train , y_test = train_test_split(digits.data , digits.target , test_size = 0.2 , random_state = 42)
```

```python
bag_clf = RandomForestClassifier(  
             n_estimators = 500 , 
             bootstrap = True , 
             n_jobs = -1
)
```

```python
bag_clf.fit(x_train , y_train)
```

```python
y_pred = bag_clf.predict(x_test)
y_pred
```

```python
accuracy = accuracy_score(y_test , y_pred)
accuracy
```
